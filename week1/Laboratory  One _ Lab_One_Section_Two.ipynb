## Exercise Set 7 – Mini Calculator

```python
# Example: Simple menu
choice = 1

if choice == 1:
    print('Addition selected')
else:
    print('Invalid option')
```

**Modified Task:** Expand this into a **calculator menu** with options for add, subtract, multiply, and divide.

```python
# TODO: Calculator menu implementation

# "choice" options from 1 to 4 

choice = 1

if choice == 1:
    print("Addition selected")

elif choice == 2:
    print("Subtraction selected")

elif choice == 3:
    print("Multiplication selected")

elif choice == 4:
    print("Division selected")

else:
    print("Invalid option")
```

**Additional Exercises:**

1. Extend the calculator so it **repeats until the user exits**.
2. Add an option to calculate the **average of two numbers**.

```python
# TODO: Extended calculator/
# 1. Extend the calculator so it **repeats until the user exits**.

while True:

    print("Calculator Menu")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Exit")

    choice = int(input("Enter your choice (1-5): "))

    if choice == 5:
        print("Calculator closed.")
        break

    elif choice == 1:
        print("Addition selected")

    elif choice == 2:
        print("Subtraction selected")

    elif choice == 3:
        print("Multiplication selected")

    elif choice == 4:
        print("Division selected")

    else:
        print("Invalid option")
```

```python
#2. Add an option to calculate the **average of two numbers**.

while True:

    print("Calculator Menu")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Exit")

    choice = int(input("Enter your choice (1-6): "))

    if choice == 5:
        print("Calculator closed.")
        break

    elif choice == 1:
        print("Addition selected")

    elif choice == 2:
        print("Subtraction selected")

    elif choice == 3:
        print("Multiplication selected")

    elif choice == 4:
        print("Division selected")

    elif choice == 5:
        num1 = float(input("Enter the first number: "))
        num2 = float(input("Enter the second number: "))

        average = (num1 + num2) / 2

        print("The average is:", average)

    else:
        print("Invalid option")
```

### 1. What is computational thinking?

Computational thinking is a way of solving problems by breaking them down into smaller and simpler steps. It helps us understand the problem and create a logical solution that can be implemented using a computer.

### 2. What is the role of algorithms in programming?

Algorithms provide a clear step-by-step process for solving a problem. They help programmers plan the logic of a program before writing the actual code.
