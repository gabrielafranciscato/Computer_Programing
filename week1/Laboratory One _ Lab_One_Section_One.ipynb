## 3. The `print()` function

`print()` displays information on the screen. It can print:

- Text (strings)
- Whole numbers (integers)
- Decimal numbers (floats)
- Multiple values at once

General form:

```python
print(<expr>, <expr>, ..., <expr>)
```

Each `<expr>` is some Python expression, for example a number or a string.


````python
# Some examples of print()

# a) Print a simple text message
print("Hello, world!")

# b) Print two pieces of text
print("Hello", "world!")

# c) Print a whole number (integer)
print(3)

# d) Print a decimal number (float)
print(3.0)

# e) Print the result of an addition
print(2 + 3)

# f) Print the result of adding two decimal numbers
print(2.0 + 3.0)
````

## 4. Getting input from the user – `input()`

The purpose of an **input statement** is to get some information from the user and store it into a **variable**.

We do this using an **assignment statement** and the built‑in `input()` function:

```python
name = input("What is your name? ")
```

- `input()` **reads a line from the user** (from the keyboard).
- It **always returns a string**.
- The variable on the left (`name`) stores whatever the user typed.


````python
# Simple example with input()
name = input("What is your name? ")
print("Hello", name, "- nice to meet you!")
````

### Converting input to numbers

Remember: `input()` returns a **string**, even if you type a number.

To do maths, you often need to convert the string to an integer using `int()`:

```python
x = int(input("Enter a number: "))
y = int(input("Enter another number: "))
print("The sum is", x + y)
```


````python
# Example: adding two numbers from the user
x = int(input("Enter a number: "))
y = int(input("Enter another number: "))
print("The sum is", x + y)
````

## 5. Order of operations (precedence)

Python follows **almost the same order of operations as mathematics**:

1. `**`  (exponent) – evaluated first  
2. `*`, `/`, `//`, `%` – evaluated next, from left to right  
3. `+`, `-` – evaluated last, from left to right  

You can always use **parentheses** `()` to control the order.

Examples:
```python
print(2 + 3 * 4)      # 3*4 happens first → 2 + 12 = 14
print((2 + 3) * 4)    # 2+3 happens first → 5 * 4 = 20
print(2 ** 3)         # 2 to the power of 3 = 8
print(7 // 3)         # integer division → 2
print(7 % 3)          # remainder of 7 divided by 3 → 1
```


````python
# Try these examples and see the results

print(2 + 3 * 4)
print((2 + 3) * 4)
print(2 ** 3)
print(7 // 3)
print(7 % 3)
````
