# COM4006 – Laboratory 3.2
## Version Control and Collaboration Using Git and GitHub

## 1.0 Lab Overview

This laboratory session introduces students to version control and collaborative software development using Git and GitHub.
Students will practise creating repositories, tracking changes, creating commits, and collaborating on projects using professional workflows.


## 2.0 Aim

To provide students with practical experience in using Git and GitHub for version control, project management, and collaborative software development workflows.


## 3.0 Learning Objectives

By the end of this laboratory session, students should be able to:

- Explain version control concepts
- Install and configure Git
- Create repositories
- Use commits to track changes
- Upload projects to GitHub
- Apply teamwork workflows


## Part 1 – Installing and Configuring Git

Download Git from: https://git-scm.com

```
git --version
```

```
git config --global user.name "Your Name"
```

```
git config --global user.email "your@email.com"
```

## Part 2 – Creating a Local Repository

```
mkdir calculator_project
```

```
cd calculator_project
```

```
git init
```

Create a Python file named calculator.py

```
print("Welcome to SmartCalc")
```

## Part 3 – Adding and Committing Files

```
git add .
```

```
git commit -m "Initial calculator program"
```

Add a function to calculator.py

```
def add(a, b):
    return a + b

print(add(4, 5))
```

```
git add .
```

```
git commit -m "Added addition function"
```

## Part 4 – Viewing Project History

```
git log
```

Observe:
- Commit IDs
- Author names
- Commit messages
- Dates


## Part 5 – Creating a GitHub Repository

Steps:
1. Create a GitHub account
2. Create a repository named calculator_project
3. Copy the repository URL


## Part 6 – Connecting Local Git to GitHub

```
git remote add origin REPOSITORY_URL
```

```
git remote -v
```

```
git push origin main
```

## Part 7 – Cloning a Repository

```
git clone REPOSITORY_URL
```

## Part 8 – Collaboration Workflow

Scenario:

Two students are developing the same calculator project.

- Student A develops menu functions
- Student B develops mathematical functions


```
git pull origin main
```

Discussion Questions:
- Why is pulling important?
- What problems occur if two students edit the same file?


## Part 9 – Introduction to Branching

```
git branch new-feature
```

```
git checkout new-feature
```

Add a new calculator feature and commit the changes.

```
git commit -m "Added new feature"
```

```
git checkout main
```

## Part 10 – Challenge Task

Extend your calculator project collaboratively.

Possible additions:
- Power function
- Percentage calculation
- Input validation
- Continuous calculator loop


## 5.0 Student Reflection Questions

1. What is version control?
2. Why are commits important?
3. What is the difference between Git and GitHub?
4. How does Git support teamwork?
5. What challenges did you face?


## 6.0 Conclusion

In this laboratory session, students practised using Git and GitHub to manage programming projects and collaborate effectively using professional software development workflows.
