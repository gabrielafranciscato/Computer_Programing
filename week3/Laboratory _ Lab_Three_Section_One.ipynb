## 1. Introduction

In Python, a **function** is a named block of code designed to perform a specific task.

Functions help programmers to:

- avoid repetition
- improve structure
- improve readability
- make programs easier to test and maintain


## 2. Defining a Function

In Python, a function is created using the `def` keyword.

**General syntax**
```python

    statement
```

A function does not run when it is defined. It runs only when it is called.


````python
def greet_user():
    print("Hello, welcome to Python programming")
````

Run the next cell to call the function.

````python
greet_user()
````

## 5. Return Statements

A **return statement** sends a value back to the point where the function was called.

This allows the result to be stored, reused, or processed further.

**General syntax**
```python
return value
```


````python
def add(a, b):
    return a + b

result = add(4, 6)
print(result)
````

````python
def square(number):
    return number * number

total = square(5) + square(3)
print(total)
````

## 6. Local and Global Variables

The **scope** of a variable refers to where that variable can be accessed.

- A **local variable** is defined inside a function and can only be used within that function.
- A **global variable** is defined outside all functions and can be accessed throughout the program.


````python
def greet():
    message = "Hello"
    print(message)

greet()
````

````python
message = "Welcome"

def greet_global():
    print(message)

greet_global()
````

````python
count = 10

def update_count():
    count = 5
    print("Inside function:", count)

update_count()
print("Outside function:", count)
````

### LEGB Rule

Python searches for variable names in this order:

1. **Local**
2. **Enclosing**
3. **Global**
4. **Built-in**


````python
x = "global"

def outer():
    x = "enclosing"
    def inner():
        x = "local"
        print("Inner:", x)
    inner()
    print("Outer:", x)

outer()
print("Global:", x)
````

Real programs usually contain several smaller functions, each responsible for one task. This is known as **modular programming**.

## 8. Calculator Program Using Four Functions

A basic calculator is a good example of modular programming. Each operation can be written as a separate function.


````python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    return a / b
````

````python
num1 = 12
num2 = 4

print("Addition:", add(num1, num2))
print("Subtraction:", subtract(num1, num2))
print("Multiplication:", multiply(num1, num2))
print("Division:", divide(num1, num2))
````
