Part 5: Modular Calculator Program
Scenario-Based Exercise: Building a Modular Calculator
You have recently been hired as a junior software developer at a small technology company
called SmartCalc Solutions.
The company wants a command-line calculator that performs:
• Addition
• Subtraction
• Multiplication
• Division
However, the senior developer has specified the following:
• Each operation must use a separate function.
• Each function must accept two parameters.
• Each function must return a value.
• The main program must handle all user interaction.
Task Requirements
Your program must:
1. Display a menu:
o 1 Addition
o 2 Subtraction
o 3 Multiplication
o 4 Division
2. Ask the user to enter two numbers.
3. Call the correct function based on the user’s choice.
4. Display the result clearly.
5. Handle invalid menu choices.
6. Prevent division by zero

```python
def add (a,b):
    return a +b 

def subtract (a,b):
    return a -b 

def multiply (a,b):
    return a*b

def divide (a,b):
    return a / b

print ("=====SmartCalc Calculator==========")
print("1. Addition")
print("2. Subtraction")
print ("3. Multiplication")
print ("4. Division")


choice = int(input("Select an operation from the list"))

if choice >= 1 and choice <=4:

    num1 = float(input("Enter the first number"))
    num2 = float(input("Enter the second number"))


    if choice ==1:
        result = add(num1,num2)
        print ("The result is:",result)


    elif choice ==2:
        result = subtract(num1,num2)
        print ("The result is",result)

    elif choice == 3:
        result =multiply (num1,num2)
        print("The result is:",result)

    elif choice == 4:
        if num2 == 0:
            print ("Cnnot divide by zero")


        elif num1 == 0:
            print("Cnnot divide by zero")

        else:
            result = divide(num1, num2)
            print("The result is:", result)

else: 
    print ("Option not listed")
```

Part 6: Challenge Task – Extended Calculator
Extend your calculator program by adding at least TWO additional operations:
Possible examples:

• Power
• Modulus
• Square root
• Percentage
• Average

```python
def power (a,b):
    return a ** b

def modulus (a,b):
    return a % b

def add (a,b):
    return a +b 

def subtract (a,b):
    return a -b 

def multiply (a,b):
    return a*b

def divide (a,b):
    return a / b

print ("=====SmartCalc Calculator==========")
print("1. Addition")
print("2. Subtraction")
print ("3. Multiplication")
print ("4. Division")
print ("5. Power")
print ("6. Modulus")


choice = int(input("Select an operation from the list"))

if choice >= 1 and choice <=6:

    num1 = float(input("Enter the first number"))
    num2 = float(input("Enter the second number"))


    if choice ==1:
        result = add(num1,num2)
        print ("The result is:",result)


    elif choice ==2:
        result = subtract(num1,num2)
        print ("The result is",result)

    elif choice == 3:
        result =multiply (num1,num2)
        print("The result is:",result)

    elif choice == 4:
        if num2 == 0:
            print ("Cnnot divide by zero")


        elif num1 == 0:
            print("Cnnot divide by zero")

        else:
            result = divide(num1, num2)
            print("The result is:", result)

    elif choice == 5:
        result = power (num1,num2)
        print ("The result is:",result)

    elif choice == 6:
        if num2 == 0:
            print ("Cannot calculate modulus by zero")
        else:
            result = modulus (num1, num2)
            print ("The result is:",result)
else: 
    print ("Option not listed")
```
