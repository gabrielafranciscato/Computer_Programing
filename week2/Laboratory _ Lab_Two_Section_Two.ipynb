## 4. Nested Decisions

A nested decision is an `if` statement inside another `if` statement.

### Worked Example: Login Check


```python
username_correct = True
password_correct = False

if username_correct:
    if password_correct:
        print("Login successful")
    else:
        print("Incorrect password")
else:
    print("Username not found")
```

## 12. FizzBuzz

FizzBuzz is a classic loop and decision exercise.

Rules:
- Print `FizzBuzz` if a number is divisible by both 3 and 5
- Print `Fizz` if divisible by 3 only
- Print `Buzz` if divisible by 5 only
- Otherwise print the number

### Step 1: Basic Counting Loop


```python
for i in range(1, 16):
    print(i)
```

### Step 2: Full FizzBuzz Solution


```python
for i in range(1, 16):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

### Practice 11
Modify the FizzBuzz loop so that it runs from 1 to 30.


```python
# TODO: modify the FizzBuzz solution here

for i in range(1, 30):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

## 13. Number Pattern, Triangle of Stars

Pattern printing helps students understand repetition and nested loops.

### Worked Example: Increasing Pattern


```python
for i in range(1, 6):
    print("*" * i)
```

### Worked Example: Nested Loop Version


```python
for i in range(1, 6):
    for j in range(i):
        print("*", end="")
    print()
```
