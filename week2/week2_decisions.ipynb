## Part 3: Multiple Conditions Using `elif`

Write a Python program that classifies a student’s score.

- Score `70` and above – **Distinction**
- Score between `50` and `69` – **Pass**
- Score below `50` – **Fail**

**Hint:** Use `if`, `elif`, and `else` statements.

```python
score = 49

if score >= 70:
    print("Disntiction")

elif score >= 50 <= 69:
    print("Pass")

elif score <= 50:
    print ("Fail")

else:
    print("Invalid score")
```

   `You can drive`

- If the person does not have a licence, print:

   `You need a licence`

- If age is below `18`, print:

   `You are too young to drive`

```python
age = int(input(" Enter your age:"))

if age >= 18:
    license = input("Do you have a driving license? (yes/no): ")

    if license =="yes":
        print("You can drive")
    else:
        print ("You need a license")

else:
    print ("You are too young to drive")
```

## Part 8: FizzBuzz Exercise

Write a program that prints numbers from `1` to `20`. Apply the following rules:

- If a number is divisible by `3`, print `Fizz`.

- If a number is divisible by `5`, print `Buzz`.

- If a number is divisible by both `3` and `5`, print `FizzBuzz`.

- Otherwise, print the number.

```python
for i in range (1,21):

    if i % 3 == 0 and i % 5 == 0:
        print ("FizzBuzz")

    elif i % 3 == 0:
        print ("Fizz")

    elif i % 5 == 0:
        print ("Buzz")

    else:
        print(i)
```
