2. Subtraction
3. Multiplication

## Part 6: Challenge Task – Continuous Calculator

This challenge encourages you to extend your calculator program.

Can you make the calculator repeat the program until the user decides to quit?

- You may need a `while` loop.
- Ask the user whether they want to continue or exit.
- Think about what condition should stop the loop.

```python
while True:

    print ('============= Menu Based Calculator ===========')
    print ("Select an option from the list bellow:")
    print ("1. Addition")
    print ("2. Subtraction")
    print ("3. Multiplication")
    print ("4. Division")

    choice = int(input(" Select the mathematic operation that you woul like to calculate:"))

    if choice == 1:
        print ("You choose addition")

        num1 = float(input("Enter first number:"))
        num2 = float(input("Enter second number:"))

        sum = (num1 + num2)

        print ("The sum is:", num1, num2)


    elif choice == 2:
        print ("You choose subtraction")

        num1 = float(input("Enter first number:"))
        num2 = float(input("Enter second number:"))

        difference= (num1 - num2)

        print ("the difference is:",difference)


    elif choice == 3:
        print ("You choose multiplication")

        num1 = float(input("Enter first number:"))
        num2 = float(input("Enter second number:"))

        product = (num1* num2)

        print ("The product is:", product)



    elif choice == 4:
        print ("You choose division")

        num1 = float(input("Enter first number:"))
        num2 = float(input("Enter second number:"))

        quotient = (num1/num2)

        print ("The quotient is:",quotient)

    else: 
        print("Invalid selection")

    continuos_calculation = input("Do you want to continue? (Yes/No)" )
    if continuos_calculation == "no":
        print ("Calculator closed.")
        break
```
