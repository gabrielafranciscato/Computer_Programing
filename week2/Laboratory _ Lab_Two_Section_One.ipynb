## 6. Operator Precedence (PEMDAS)

When an expression contains more than one operator, Python follows an order of evaluation:
1. Parentheses `()`
2. Exponents `**`
3. Multiplication, division, integer division, modulus `* / // %`
4. Addition and subtraction `+ -`

If operators have the same level of priority, Python works from left to right.

```python
# Worked Example 1: Operator precedence
print(5 + 2 * 3)
print((5 + 2) * 3)
print(10 - 4 + 2)
print(20 / 5 * 2)
print(2 ** 3 * 2)
print(2 ** (3 * 2))
```
