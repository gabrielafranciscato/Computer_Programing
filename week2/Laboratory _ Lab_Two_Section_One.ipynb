## 2. Floats (`float`)

A **float** represents a number with a decimal point. Floats are commonly used for measurements, averages, and prices.

```
# Worked Example 1: Mean of three numbers
score1 = 12.5
score2 = 8.75
score3 = 10.0

mean = (score1 + score2 + score3) / 3
print("Mean:", mean)
```

```
# Worked Example 2: round() and abs()
price = 9.876
temperature = -3.5

print("Rounded price:", round(price, 2))
print("Absolute value:", abs(temperature))
```

## 3. Strings (`str`)

```
# Worked Example 1: Creating and printing strings
first_name = "Douglas"
course_name = "Computer Programming"

message = "Welcome " + first_name + " to the " + course_name + " course"
print(message)
```

## 4. Booleans (`bool`)

A **Boolean** has only two possible values: `True` or `False`. Booleans are often produced by comparisons and are used in decision-making.

```
# Worked Example 1: Comparison operators
x = 5

print(x == 5)
print(x == 3)
print(10 != 5)
print(8 > 5)
print(3 < 1)
```

```
# Worked Example 2: Logical operators
age = 20
is_student = True
has_id = False
has_pass = True
logged_in = False

print(age > 18 and is_student)
print(has_id or has_pass)
print(not logged_in)
```

## 5. Type Casting

Casting means converting a value from one data type to another. Common constructor functions are:
- `int()`
- `float()`
- `str()`

```
# Worked Example 1: Casting to int, float, and str
x = int(1)
y = int(2.8)
z = int("3")

a = float(1)
b = float(2.8)
c = float("4.2")

m = str(2)
n = str(3.0)

print(x, y, z)
print(a, b, c)
print(m, n)
```

## 6. Operator Precedence (PEMDAS)

When an expression contains more than one operator, Python follows an order of evaluation:
1. Parentheses `()`
2. Exponents `**`
3. Multiplication, division, integer division, modulus `* / // %`
4. Addition and subtraction `+ -`

If operators have the same level of priority, Python works from left to right.

```
# Worked Example 1: Operator precedence
print(5 + 2 * 3)
print((5 + 2) * 3)
print(10 - 4 + 2)
print(20 / 5 * 2)
print(2 ** 3 * 2)
print(2 ** (3 * 2))
```

```
# Worked Example 2: Predict the output before running
print(12 / 3 + 2 * 4)
print((12 / 3 + 2) * 4)
```
